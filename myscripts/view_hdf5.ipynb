# View Robomimic HDF5

Example: open `stack_d1.hdf5`, count episodes under `data/`, and list field names for one episode.

Conda env: Robosuite (Python 3.10.20)

```python
import h5py

HDF5_PATH = "/home/khw/Projects/equidiff/data/robomimic/datasets/stack_d1/stack_d1.hdf5"

with h5py.File(HDF5_PATH, "r") as f:
    data = f["data"]
    demo_keys = sorted(data.keys(), key=lambda k: int(k.split("_")[1]))
    n_demos = len(demo_keys)

    first = demo_keys[0]
    grp = data[first]
    actions = grp["actions"]
    n_timesteps = int(actions.shape[0])

    print("file:", HDF5_PATH)
    print("number of episodes (demo groups under data/):", n_demos)
    print(f"example {first!r} — timesteps (rows along time axis in actions):", n_timesteps)
    print("episode group field names:", list(grp.keys()))
    print("obs/* field names:", list(grp["obs"].keys()))
    print("actions shape, dtype:", actions.shape, actions.dtype)
```
